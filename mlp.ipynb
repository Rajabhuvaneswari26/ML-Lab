```python

import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

def checkError(predicted):
    expected = np.array([[0],[1],[1],[0]])
    return np.any(np.abs(expected - predicted) > 0.001)

inputs = np.array([[0,0],[0,1],[1,0],[1,1]])
expected_output = np.array([[0],[1],[1],[0]])

inputLayerNeurons = int(input("Enter no of input layer: "))
hiddenLayerNeurons = int(input("Enter no of hidden layer: "))
outputLayerNeurons = int(input("Enter no of output layer: "))

hidden_weights = []
for i in range(inputLayerNeurons):
    row = []
    for j in range(hiddenLayerNeurons):
        row.append(float(input("Enter weight: ")))
    hidden_weights.append(row)

output_weights = []
for i in range(hiddenLayerNeurons):
    row = []
    for j in range(outputLayerNeurons):
        row.append(float(input("Enter weight: ")))
    output_weights.append(row)

hidden_bias = []
for i in range(hiddenLayerNeurons):
    hidden_bias.append(float(input("Enter hidden bias: ")))

output_bias = []
for i in range(outputLayerNeurons):
    output_bias.append(float(input("Enter output bias: ")))

hidden_weights = np.array(hidden_weights)
output_weights = np.array(output_weights)
hidden_bias = np.array([hidden_bias])
output_bias = np.array([output_bias])

print("\nInitial hidden weights:")
print(hidden_weights)

print("Initial hidden biases:")
print(hidden_bias)

print("Initial output weights:")
print(output_weights)

print("Initial output biases:")
print(output_bias)

lr = 0.1
epoch = 0
predicted_output = np.zeros((4,1))

while checkError(predicted_output):

    epoch += 1

    hidden_activation = np.dot(inputs, hidden_weights) + hidden_bias
    hidden_output = sigmoid(hidden_activation)

    output_activation = np.dot(hidden_output, output_weights) + output_bias
    predicted_output = sigmoid(output_activation)

    error = expected_output - predicted_output

    d_output = error * sigmoid_derivative(predicted_output)

    error_hidden = d_output.dot(output_weights.T)
    d_hidden = error_hidden * sigmoid_derivative(hidden_output)

    output_weights += hidden_output.T.dot(d_output) * lr
    output_bias += np.sum(d_output, axis=0, keepdims=True) * lr

    hidden_weights += inputs.T.dot(d_hidden) * lr
    hidden_bias += np.sum(d_hidden, axis=0, keepdims=True) * lr

print("\nFinal hidden weights:")
print(hidden_weights)

print("Final hidden bias:")
print(hidden_bias)

print("Final output weights:")
print(output_weights)

print("Final output bias:")
print(output_bias)

print("\nOutput from neural network:")
print(predicted_output)

print("\nNo of epochs:")
print(epoch)
```
