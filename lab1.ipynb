```python
# LAB EXERCISE 1
import pandas as pd
col_list = ["id", "first", "last", "gender", "Marks", "selected"]
df = pd.read_csv("sample.csv", usecols=col_list)

print("SAMPLE DATASET")
print(df)
mean1 = df['Marks'].mean()
sum1 = df['Marks'].sum()
max1 = df['Marks'].max()
min1 = df['Marks'].min()
count1 = df['Marks'].count()
median1 = df['Marks'].median()
sd1 = df['Marks'].std()
var1 = df['Marks'].var()

print("\nSTATISTICAL SUMMARY")

print('Mean Marks\n' + str(mean1))

print('Sum of the Marks\n' + str(sum1))

print('Maximum of the Marks\n' + str(max1))

print('Minimum of the Marks\n' + str(min1))

print('Count of the Marks\n' + str(count1))

print('Median of the Marks\n' + str(median1))

print('Standard Deviation of the Marks\n' + str(sd1))

print('Variance of the Marks\n' + str(var1))

print('End of Summary')
print("\n\nDATASET EXPLORATION")

print("\nDataset:")
print(df)

print("\nShape of the Dataset:")
print(df.shape)

print("\nFirst 5 Rows:")
print(df.head(5))

print("\nDescriptive Statistics:")
print(df.describe())

from scipy import stats
data = [
    1, 2, 3, 4, 5, 6,
    8, 8, 8, 8,
    9,
    10, 10, 10,
    12, 14,
    18, 18, 18,
    22, 39, 44,
    55, 55, 55, 55,
    66, 78, 79, 88
]

print("\nDetails of the Data")
print(stats.describe(data))
print("\nThe Cumulative Frequency of the Data")
print(stats.cumfreq(data))

print("\nThe Geometric Mean of the Data")
print(stats.gmean(data))

print("\nThe Harmonic Mean of the Data")
print(stats.hmean(data))

print("\nThe IQR of the Data")
print(stats.iqr(data))

print("\nThe Z-score of the Data")
print(stats.zscore(data))

print("\nThe Skewness of the Data")
print(stats.skew(data))

print("\nThe Kurtosis of the Data")
print(stats.kurtosis(data))
data = [1, 2, 3, 6, 8, 10]

data1 = [2, 3, 4, 5, 9, 12]
print("\nThe Spearman Correlation of the Data")
print(stats.spearmanr(data, data1))
```
