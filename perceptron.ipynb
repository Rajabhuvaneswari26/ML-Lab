```python
import numpy as np

class Perceptron:

    def __init__(self, lr=0.2, epochs=4):
        self.W = np.array([0.3, -0.2])
        self.lr = lr
        self.epochs = epochs

    def activation(self, x):
        return 1 if x >= 0 else 0

    def predict(self, x, theta):
        z = np.dot(self.W, x) - theta
        return self.activation(z)

    def fit(self, X, d, theta):
        for epoch in range(self.epochs):
            print("Epoch:", epoch + 1)

            for i in range(len(X)):
                y = self.predict(X[i], theta)
                error = d[i] - y
                self.W += self.lr * error * X[i]

                print("Input:", X[i], "Weight:", self.W)

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

d = np.array([0, 0, 0, 1])

p = Perceptron()

p.fit(X, d, 0.4)

print("Final Weights:", p.W)
```
