```python
# Program: Plot of y = x²

import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 1, 10)
y = x ** 2

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("y = x²")
plt.title("Graph of y = x²")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,5,100)

plt.subplot(211)
plt.plot(x, x**2)
plt.title("y = x²")

plt.subplot(212)
plt.plot(x, x**3)
plt.title("y = x³")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,5,100)

fig, a = plt.subplots(2,2)

a[0][0].plot(x, x**2)
a[0][0].set_title("x²")

a[1][1].plot(x, x**3)
a[1][1].set_title("x³")

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("iris")

sns.histplot(df["sepal_width"])
plt.title("Distribution Plot")
plt.show()
```
