```python

import numpy as np

print("Creating a NumPy Array")
x = np.array([1, 2, 3, 4, 5])

print("Array x:")
print(x)

print("\nData Type of x:")
print(type(x))

print("\nShape of x:")
print(x.shape)

print("\nArray using arange():")
arr1 = np.arange(1, 10, 5)
print(arr1)

print("\nArray using linspace():")
arr2 = np.linspace(1, 10, 5)
print(arr2)


print("\nCreating another array:")
x = np.array([1, 2, 3, 4, 5, 6])
print(x)
```

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5, 6])
print("1. Reshape (2 x 3):")
print(x.reshape(2, 3))

print("\n2. Size:")
print(x.size)


print("\n3. Shape:")
print(x.shape)

print("\n4. Data Type:")
print(x.dtype)

print("\n5. Item Size:")
print(x.itemsize)

print("\n6. Zeros Array:")
print(np.zeros((4, 5)))

print("\n7. Ones Array:")
print(np.ones((4, 5)))

print("\n8. Random Array:")
a = np.random.random((4, 5))
print(a)

print("\n9. Random Integers:")
x = np.random.randint(0, 10, 5)
print(x)

print("\n10. Shuffled Array:")
np.random.shuffle(x)
print(x)


print("\n11. Random Choice from Array a:")
print(np.random.choice(a.flatten()))
```

```python
import numpy as np

a = np.arange(15)

print("Original Array:")
print(a)

print("\nSlicing a[3:9:2]:")
print(a[3:9:2])

print("\nSlicing a[3:]:")
print(a[3:])

print("\nSlicing a[3:8]:")
print(a[3:8])

x = np.array([1, 2, 3, 4, 5, 6])
y = np.array([5, 6, 7, 8, 9, 10])

print("\nArray x:")
print(x)

print("\nArray y:")
print(y)

print("\nAddition:")
print(x + y)

print("\nSubtraction:")
print(x - y)

print("\nMultiplication:")
print(x * y)

print("\nDivision:")
print(x / y)

print("\nMean:", np.mean(x))
print("Median:", np.median(x))
print("Maximum:", np.max(x))
print("Minimum:", np.min(x))
print("Sum:", x.sum())
```

```python
import numpy as np
x = np.array([[1, 2, 3],
              [2, 2, 2]])

print("2D Array:")
print(x)


print("\nRow-wise Sum:")
print(x.sum(axis=1))


print("\nColumn-wise Sum:")
print(x.sum(axis=0))

a = np.array([4, 5, 6])
b = np.array([10, 10, 10])

print("\nVector Multiplication:")
print(a * b)

m1 = np.array([[2, 3],
               [6, 7]])

m2 = np.array([[5, 6],
               [8, 7]])

print("\nMatrix Addition:")
print(np.add(m1, m2))
```

```python
# PROGRAM: Ravel and Matrix Operations

import numpy as np

a = np.arange(15).reshape(3,5)

print("Original Matrix:")
print(a)

print("\nFlattened Array:")
print(a.ravel())

x = np.array([[1,2],[3,4]])
y = np.array([[5,6],[7,8]])

print("\nAddition:")
print(np.add(x,y))


print("\nSubtraction:")
print(np.subtract(x,y))

print("\nMultiplication:")
print(np.matmul(x,y))


print("\nDivision:")
print(np.divide(x,y))

print("\nDeterminant of x:")
print(np.linalg.det(x))
```
