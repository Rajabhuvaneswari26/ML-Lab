```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Iris.csv")

data = df.iloc[1:20]

# Scatter Plot
data.plot.scatter(x='SepalLengthCm', y='SepalWidthCm')
plt.show()

# Histogram
data.hist(column='SepalLengthCm', by='Species')
plt.show()

# Box Plot
data.boxplot(column='SepalLengthCm', by='Species')
plt.show()
```
