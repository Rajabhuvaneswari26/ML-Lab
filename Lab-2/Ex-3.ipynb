```python
import pandas as pd
df = pd.DataFrame({
                'm1':[50,'A',60,'A',80],
                'm2':[60,'A','60','A',80],
                'm3':[50,70,'A','A',60],
                'm4':[60,'A','A','A',60],
                'm5':['A','A','A',10,20]
                })

df = df.apply(pd.to_numeric,errors='coerce')

print(df)

print('Dataframe with NaN\n\n\n')

df['m5']=df['m5'].fillna(0)
print(df)
print('Making m5 NaN as 0 using fillna() function\n\n\n\n')

df1 = df.copy()
df1['m2'].fillna(df1['m2'].mean(),inplace=True)
print(df1)
print('Making m5 NaN as mean using fillna() function\n\n\n\n')

df2 = df.copy()
df1['m3'].fillna(df1['m2'].median(),inplace=True)
print(df2)
print('Making m5 NaN as median using fillna() function\n\n\n\n')


df=df.dropna(axis=1)
print(df)
print('Dropping all columns having NaN\n\n\n\n')
```

```python
import pandas as pd

# Create DataFrame
df = pd.DataFrame({
    'm1': [50, 'A', 60, 'A', 80],
    'm2': [60, 'A', '60', 'A', 80],
    'm3': [50, 70, 'A', 'A', 60],
    'm4': [60, 'A', 'A', 'A', 60],
    'm5': ['A', 'A', 'A', 10, 20]
})

# Convert non-numeric values to NaN
df = df.apply(pd.to_numeric, errors='coerce')

print(df)
print("Dataframe with NaN\n\n")

# Replace NaN in m5 with 0
df['m5'] = df['m5'].fillna(0)
print(df)
print("Making m5 NaN as 0 using fillna() function\n\n")

# Replace NaN in m2 with Mean
df1 = df.copy()
df1['m2'] = df1['m2'].fillna(df1['m2'].mean())
print(df1)
print("Making m2 NaN as Mean using fillna() function\n\n")

# Replace NaN in m3 with Median
df2 = df.copy()
df2['m3'] = df2['m3'].fillna(df2['m3'].median())
print(df2)
print("Making m3 NaN as Median using fillna() function\n\n")

# Drop columns containing NaN
df3 = df.copy()
df3 = df3.dropna(axis=1)
print(df3)
print("Dropping all columns having NaN")
```
