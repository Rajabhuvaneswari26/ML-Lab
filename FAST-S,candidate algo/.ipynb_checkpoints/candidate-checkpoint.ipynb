```python
import csv
import copy

a = []

print("\nTraining Dataset\n")

with open('Find-S.csv', 'r') as csvFile:
    reader = csv.reader(csvFile)
    for row in reader:
        a.append(row)
        print(row)

num_attrs = len(a[0]) - 1

# Initialize S and G
S = ['0'] * num_attrs
G = [['?' for i in range(num_attrs)] for j in range(num_attrs)]

print("\nInitial Specific Hypothesis:")
print(S)

print("\nInitial General Hypothesis:")
print(G)

for i in range(len(a)):

    if a[i][num_attrs] == "Yes":

        for j in range(num_attrs):
            if S[j] == '0':
                S[j] = a[i][j]
            elif S[j] != a[i][j]:
                S[j] = '?'
                G[j][j] = '?'

    else:

        for j in range(num_attrs):
            if S[j] != a[i][j] and S[j] != '?':
                G[j][j] = S[j]
            else:
                G[j][j] = '?'

    print("\nTraining Instance", i + 1)
    print("Specific Hypothesis:", S)
    print("General Hypothesis:")
    for g in G:
        print(g)

# Remove redundant hypotheses
G = [g for g in G if g != ['?'] * num_attrs]

print("\nFinal Specific Hypothesis:")
print(S)

print("\nFinal General Hypothesis:")
for g in G:
    print(g)
```
