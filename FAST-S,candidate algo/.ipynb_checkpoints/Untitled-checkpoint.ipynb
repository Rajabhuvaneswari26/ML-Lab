```python
import pandas as pd

# Training data
data = {
    "Sky": ["Sunny", "Sunny", "Rainy", "Sunny"],
    "AirTemp": ["Warm", "Warm", "Cold", "Warm"],
    "Humidity": ["Normal", "High", "High", "Normal"],
    "Wind": ["Strong", "Strong", "Strong", "Weak"],
    "EnjoySport": ["Yes", "Yes", "No", "Yes"]
}

df = pd.DataFrame(data)

# Initialize hypothesis
hypothesis = None

# Find-S Algorithm
step = 1
for _, row in df.iterrows():

    if row["EnjoySport"] == "Yes":

        if hypothesis is None:
            hypothesis = row.iloc[:-1].tolist()   # First positive example
        else:
            for i in range(len(hypothesis)):
                if hypothesis[i] != row.iloc[i]:
                    hypothesis[i] = "?"

        print(f"Step {step}: {hypothesis}")
        step += 1

print("\nFinal Hypothesis:", hypothesis)
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Dataset
X = np.array([[1],[2],[3],[4],[5]])
y = np.array([2,4,6,8,10])

# Create model
model = LinearRegression()

# Train model
model.fit(X, y)

# Predict
prediction = model.predict([[6]])

# Print outputs
print("Prediction:", prediction)
print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)
print("R2 Score:", model.score(X, y))
```
